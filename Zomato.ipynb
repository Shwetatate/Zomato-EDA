```python
combine_df= pd.merge(df,df_country,on='Country Code', how ='left')
```

```python
combine_df.head(3)
```

```python
combine_df.columns
```

```python
country_names=combine_df.Country.value_counts().index
```

```python
print(country_names)
```

```python
country_val=combine_df.Country.value_counts()
```

```python
print(country_val)
```

```python
##pie chart -
plt.pie(country_val,labels=country_names)
```

```python
#pie chart of top 3 countries of zomato
plt.pie(country_val[:3],labels=country_names[:3])
```

```python
#include percentage into the pie chart
plt.pie(country_val[:3],labels=country_names[:3],autopct="%1.2f%%")
```

#observation :there are 94.39 transcation done in india which is the maximum transcation carried out and then usa and uk

```python
combine_df.columns
```

```python
#we have grouped all the three columns
combine_df.groupby(['Aggregate rating','Rating color','Rating text']).size()
```

```python
#we have remaned the 0th column as Rating count

ratings=combine_df.groupby(['Aggregate rating','Rating color','Rating text']).size().reset_index().rename(columns={0:"Rating count"})
```

observation : 1. 4.5 ----4.9 then excellent
              2.4.0 -----4.4 then the very good
              3.3.5 --4.0 then the good
              4.2.5 --3.4 then the rating is average
              5.1.8---to 2.4 then its is poor

```python
print(ratings.head())
```

```python
import matplotlib
matplotlib.rcParams['figure.figsize']=(10,6) # to increse the data
sns.barplot(x="Aggregate rating",y="Rating count",hue = 'Rating color',data= ratings)
```

observation : more than 2000 ppl has not rated
